# UMAP Visualization for RAG System

This notebook demonstrates how to visualize document embeddings and query relationships using UMAP dimensionality reduction.

## What you'll see:
- 🔵 Document chunks in 2D space
- 🔴 Original query
- 🟠 Sub-queries (different perspectives)
- 🟢 HyDE (Hypothetical Document)
- 🟣 Step-back query
- 🔷 Retrieved documents

## 1. Setup and Imports

```python
import os
import numpy as np
import plotly.graph_objects as go
import chromadb
from chromadb.config import Settings
from sentence_transformers import SentenceTransformer
import umap
from pathlib import Path
import sys

# Add parent directory to path to import config
PROJECT_ROOT = Path(__file__).parent

EMBEDDING_MODEL = os.getenv("EMBEDDING_MODEL", "all-MiniLM-L6-v2")
VECTOR_DB_PATH = PROJECT_ROOT / "vector_db"

print("✅ All imports successful")
print(f"📊 Embedding Model: {EMBEDDING_MODEL}")
print(f"💾 Vector DB Path: {VECTOR_DB_PATH}")
```

## 2. Load Your Collection

```python
# ===== CONFIGURATION =====
COLLECTION_NAME = "rtyui"  # Change this to your collection name
QUERY = "What is this document about?"  # Change this to your query

# Initialize
model = SentenceTransformer(EMBEDDING_MODEL)
client = chromadb.PersistentClient(
    path=str(VECTOR_DB_PATH),
    settings=Settings(anonymized_telemetry=False)
)

# List all collections
collections = client.list_collections()
print("\n📚 Available Collections:")
for col in collections:
    print(f"  - {col.name}")

# Get your collection
try:
    collection = client.get_collection(name=COLLECTION_NAME)
    count = collection.count()
    print(f"\n✅ Loaded collection '{COLLECTION_NAME}' with {count} documents")
except Exception as e:
    print(f"\n❌ Error loading collection: {e}")
    print("Please update COLLECTION_NAME variable above")
```

## 3. Retrieve Document Embeddings

```python
# Try multiple methods to get embeddings (workaround for ChromaDB bug)
doc_embeddings = None
doc_texts = None

try:
    # Method 1: Standard get
    print("Trying Method 1: collection.get(include=[...])")
    data = collection.get(include=["embeddings", "documents"])
    embeddings_list = data.get("embeddings", [])
    doc_texts = data.get("documents", [])
    if embeddings_list and len(embeddings_list) > 0:
        doc_embeddings = np.array(embeddings_list)
        print(f"✅ Method 1 worked: Got {len(doc_embeddings)} embeddings")
except Exception as e:
    print(f"⚠️ Method 1 failed: {e}")
    
    try:
        # Method 2: Get all data
        print("Trying Method 2: collection.get()")
        all_data = collection.get()
        embeddings_list = all_data.get("embeddings", [])
        doc_texts = all_data.get("documents", [])
        if embeddings_list and len(embeddings_list) > 0:
            doc_embeddings = np.array(embeddings_list)
            print(f"✅ Method 2 worked: Got {len(doc_embeddings)} embeddings")
    except Exception as e2:
        print(f"⚠️ Method 2 failed: {e2}")
        
        try:
            # Method 3: Use query
            print("Trying Method 3: collection.query()")
            query_emb = model.encode([QUERY])
            query_result = collection.query(
                query_embeddings=query_emb.tolist(),
                n_results=min(100, collection.count()),
                include=["embeddings", "documents"]
            )
            embeddings_list = query_result["embeddings"][0] if query_result["embeddings"] else []
            doc_texts = query_result["documents"][0] if query_result.get("documents") else []
            if embeddings_list and len(embeddings_list) > 0:
                doc_embeddings = np.array(embeddings_list)
                print(f"✅ Method 3 worked: Got {len(doc_embeddings)} embeddings")
        except Exception as e3:
            print(f"❌ All methods failed: {e3}")

if doc_embeddings is not None:
    print(f"\n📊 Successfully loaded {len(doc_embeddings)} document embeddings")
    print(f"📝 Embedding dimension: {doc_embeddings.shape[1]}")
else:
    print("\n❌ Failed to load embeddings")
```

## 4. Generate Query Embeddings

```python
# Create different query variations
queries = {
    "Original": QUERY,
    "Sub-Q1": "What are the main concepts?",
    "Sub-Q2": "What are the key findings?",
    "Sub-Q3": "What methodology is used?",
    "HyDE": "This document discusses advanced techniques in machine learning and presents experimental results showing improvements over baseline methods.",
    "Step-Back": "What are the fundamental principles discussed?"
}

print("🔍 Generating embeddings for queries:\n")
all_queries = []
query_names = []
for name, query in queries.items():
    all_queries.append(query)
    query_names.append(name)
    print(f"  {name}: {query[:60]}...")

# Encode all queries
query_embeddings = model.encode(all_queries)
print(f"\n✅ Generated {len(query_embeddings)} query embeddings")
```

## 5. Apply UMAP Dimensionality Reduction

```python
# Combine all embeddings
all_embeddings = np.vstack([doc_embeddings, query_embeddings])
print(f"📊 Combined embeddings shape: {all_embeddings.shape}")

# Apply UMAP
n_neighbors = min(15, len(all_embeddings) - 1)
print(f"\n🔧 UMAP parameters:")
print(f"  - n_neighbors: {n_neighbors}")
print(f"  - min_dist: 0.1")
print(f"  - metric: cosine")

reducer = umap.UMAP(
    n_neighbors=n_neighbors,
    min_dist=0.1,
    random_state=42,
    metric="cosine"
)

print("\n⏳ Reducing dimensions (this may take a moment)...")
all_2d = reducer.fit_transform(all_embeddings)
print("✅ UMAP reduction complete!")

# Split back into documents and queries
doc_2d = all_2d[:len(doc_embeddings)]
query_2d = all_2d[len(doc_embeddings):]

print(f"\n📍 2D Coordinates:")
print(f"  - Documents: {doc_2d.shape}")
print(f"  - Queries: {query_2d.shape}")
```

## 6. Create Interactive Visualization

```python
# Create figure
fig = go.Figure()

# 1. Add document chunks (light blue)
hover_texts = []
for i, text in enumerate(doc_texts):
    preview = str(text)[:80] + "..." if len(str(text)) > 80 else str(text)
    hover_texts.append(f"Doc {i}: {preview}")

fig.add_trace(go.Scatter(
    x=doc_2d[:, 0],
    y=doc_2d[:, 1],
    mode='markers',
    marker=dict(size=6, color='lightblue', opacity=0.6),
    text=hover_texts,
    hoverinfo='text',
    name=f'Documents ({len(doc_2d)})'
))

# 2. Add queries with different colors
colors = ['red', 'orange', 'orange', 'orange', 'green', 'purple']
symbols = ['star', 'circle', 'circle', 'circle', 'circle', 'circle']
sizes = [18, 12, 12, 12, 12, 12]

for i, name in enumerate(query_names):
    fig.add_trace(go.Scatter(
        x=[query_2d[i, 0]],
        y=[query_2d[i, 1]],
        mode='markers+text',
        marker=dict(
            size=sizes[i],
            color=colors[i],
            symbol=symbols[i],
            line=dict(width=2, color='black')
        ),
        text=name,
        textposition='top center',
        textfont=dict(size=10, color='black'),
        hovertext=f"<b>{name}</b><br>{all_queries[i][:150]}",
        hoverinfo='text',
        name=name
    ))

# 3. Update layout
fig.update_layout(
    title=dict(
        text=f"UMAP Embedding Space Visualization<br><sub>Query: '{QUERY[:60]}...'</sub>",
        x=0.5,
        xanchor='center'
    ),
    xaxis_title="UMAP Dimension 1",
    yaxis_title="UMAP Dimension 2",
    hovermode="closest",
    height=700,
    width=1000,
    showlegend=True,
    legend=dict(
        x=0.01,
        y=0.99,
        bgcolor="rgba(255,255,255,0.9)",
        bordercolor="black",
        borderwidth=1
    ),
    template="plotly_white"
)

fig.show()
print("\n✅ Visualization complete! Hover over points for details.")
```

## 7. Add Retrieved Documents (Optional)

```python
# Retrieve top documents for the query
from retriever import Retriever

retriever = Retriever()
retrieved_data = retriever.retrieve(QUERY, COLLECTION_NAME, top_k=5)
retrieved_docs = retrieved_data['documents']

print(f"📚 Retrieved {len(retrieved_docs)} documents")

# Encode retrieved docs
if retrieved_docs and len(retrieved_docs) > 0:
    retrieved_embeddings = model.encode(retrieved_docs)
    
    # Transform to UMAP space
    retrieved_2d = reducer.transform(retrieved_embeddings)
    
    # Add to existing figure
    retrieved_hover = [str(doc)[:100] for doc in retrieved_docs]
    
    fig.add_trace(go.Scatter(
        x=retrieved_2d[:, 0],
        y=retrieved_2d[:, 1],
        mode='markers',
        marker=dict(
            size=12,
            color='darkblue',
            symbol='diamond',
            line=dict(width=2, color='black')
        ),
        text=retrieved_hover,
        hoverinfo='text',
        name=f'Retrieved ({len(retrieved_2d)})'
    ))
    
    fig.show()
    print("\n✅ Added retrieved documents to visualization!")
else:
    print("\n⚠️ No documents retrieved")
```

## 8. Analysis and Statistics

```python
# Calculate distances between query and documents
from scipy.spatial.distance import cosine

print("📊 Distance Analysis\n")

# Original query vs documents
original_query_emb = query_embeddings[0]
distances = []
for doc_emb in doc_embeddings:
    dist = cosine(original_query_emb, doc_emb)
    distances.append(dist)

distances = np.array(distances)

print(f"Query to Documents:")
print(f"  - Mean distance: {distances.mean():.4f}")
print(f"  - Min distance: {distances.min():.4f}")
print(f"  - Max distance: {distances.max():.4f}")
print(f"  - Std deviation: {distances.std():.4f}")

# Find closest documents
closest_indices = np.argsort(distances)[:5]
print(f"\n🎯 Top 5 Closest Documents:")
for i, idx in enumerate(closest_indices):
    print(f"  {i+1}. Distance: {distances[idx]:.4f}")
    print(f"     Text: {doc_texts[idx][:80]}...\n")
```

## 9. Export Visualization

```python
# Save as HTML
output_file = f"umap_visualization_{COLLECTION_NAME}.html"
fig.write_html(output_file)
print(f"\n💾 Visualization saved to: {output_file}")

# Save as PNG (requires kaleido)
# fig.write_image("umap_visualization.png", width=1200, height=800)
```

## 10. Interpretation Guide

### 🎨 Color Legend:
- **🔵 Light Blue Dots**: All document chunks in your collection
- **🔴 Red Star**: Your original query
- **🟠 Orange Circles**: Sub-queries (different perspectives on the question)
- **🟢 Green Circle**: HyDE (Hypothetical Document Embedding - what an answer might look like)
- **🟣 Purple Circle**: Step-back query (broader conceptual question)
- **🔷 Dark Blue Diamonds**: Retrieved documents (most relevant to your query)

### 📖 How to Read the Map:
1. **Proximity = Similarity**: Points close together are semantically similar
2. **Clusters**: Groups of documents about similar topics
3. **Good Retrieval**: Retrieved documents (diamonds) should be near your queries
4. **HyDE Effectiveness**: Green circle should land in document-dense regions
5. **Query Coverage**: Spread-out query variations cover different aspects of your question

### ✅ Quality Indicators:
- **Good**: Retrieved docs cluster near queries
- **Good**: Sub-queries explore different semantic regions
- **Good**: HyDE lands in relevant document areas
- **Bad**: Retrieved docs far from all queries
- **Bad**: All queries collapse to same point

### 🔧 Troubleshooting:
- **All points in one blob**: Try different UMAP parameters (n_neighbors, min_dist)
- **Too spread out**: Increase n_neighbors or min_dist
- **Queries too close**: Queries might be too similar, diversify them
- **No clear clusters**: Document set might be too homogeneous or too diverse
